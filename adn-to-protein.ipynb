## Ejercicio 1. Replicación del ADN

### 1. Considera la siguiente secuencia de ADN: 5’ – ATG CCG TTA GCT – 3’ / 3’ – TAC GGC AAT CGA – 5’.

### 2. Realiza una ronda de replicación:
- #### Identifica las nuevas hebras que se formarán.
La replicación del ADN es semiconservativa, lo que significa que cada nueva molécula de ADN estará compuesta por una hebra original y una hebra recién sintetizada. Dada la secuencia de ADN proporcionada:
- **Hebra parental 1:** 5’ – ATG CCG TTA GCT – 3’
    - **Nueva hebra complementaria 1:** 3’ – TAC GGC AAT CGA – 5’

- **Hebra parental 2:** 3’ – TAC GGC AAT CGA – 5’
    - **Nueva hebra complementaria 2:** 5’ – ATG CCG TTA GCT – 3’

Se acaban obteniendo dos moléculas de ADN idénticas a la original, cada una con una hebra parental y una hebra recién sintetizada.

- #### Indica cuál sería la función de las enzimas helicasa, primasa, ADN polimerasa y ligasa en este proceso.
    - **Helicasa:** Rompe los puentes de hidrógeno que unen las bases nitrogenadas de las dos hebras del ADN, separando las hebras parentales y permitiendo que sirvan de molde.
    - **Primasa:** Sintetiza pequeños fragmentos de ARN que sirven como cebadores. Esto proporciona el punto de partida para que la ADN polimerasa pueda iniciar la síntesis de la nueva hebra, ya que esta enzima no puede comenzar una cadena desde cero.
    - **ADN polimerasa:** Sintetiza la nueva hebra de ADN añadiendo nucleótidos complementarios a la hebra molde (a partir del cebador) y cuenta con mecanismos de corrección de errores para asegurar la fidelidad de la copia.
    - **Ligasa:** Une los fragmentos de ADN recién sintetizados (fragmentos de Okazaki) en la hebra rezagada, formando una hebra continua.

### 3. Reflexiona: ¿qué ocurriría si la ADN polimerasa cometiera un error en una base y no se corrigiera?
Se producirá una mutación genética permanente en la secuencia de ADN de la célula que daría lugar a una reacción en cadena con las siguientes consecuencias:
- Cuando la célula se divida, la base errónea servirá de molde para las siguientes replicaciones, heredándose la mutación a las células hijas.
- Dependiendo de la ubicación y el tipo de mutación, el cambio de base puede ser silencioso, o alterar la secuencia de aminoácidos de una proteína, modificando o anulando su función.
- Si la mutación afecta a genes clave, puede provocar disfunciones celulares graves o contribuir al desarrollo de enfermedades como el cáncer.

### Extension con Biopython:

## Ejercicio  2. Transcripción del ADN a ARN


### 1. Usa la siguiente secuencia de ADN: 5’ – ATG CCT GAA TGC – 3’ / 3’ – TAC GGA CTT ACG – 5’

### 2. Identifica cuál es la cadena molde.
La ARN polimerasa solo puede sintetizar la nueva cadena de ARN en dirección $5' \to 3'$, por lo que necesita leer una hebra de ADN antiparalela en sentido $3' \to 5'$.Por tanto:
- **Cadena molde:** 3’ – TAC GGA CTT ACG – 5’ (es la hebra que lee físicamente la ARN polimerasa para sintetizar el transcrito).

### 3. Obtén el transcrito de ARN correspondiente (recuerda que el ARN se sintetiza en dirección 5’ → 3’ y utiliza uracilo en lugar de timina).
La ARN polimerasa lee la cadena molde (3’ – TAC GGA CTT ACG – 5’) y sintetiza la molécula de ARN en dirección $5’ \to 3’$ añadiendo las bases complementarias ($A \to U$, $T \to A$, $C \to G$, $G \to C$). (Esto equivale a tomar la cadena codificante, la complementaria a la cadena molde, y sustituir la timina por uracilo).
- **Transcrito de ARN:** 5’ – AUG CCU GAA UGC – 3’

### 4. Explica cuál sería la región promotora y cuál la región codificante.
- **Región promotora:** Es una secuencia reguladora del ADN situada antes del inicio del gen (hacia el extremo $5'$ de la hebra codificante). Su función es servir de sitio de reconocimiento y unión para la ARN polimerasa, indicándole dónde comenzar la transcripción y en qué dirección transcribir; esta región no se transcribe ni se traduce a proteína. (En el ejercicio, se encontraría situada justo antes del triplete inicial ATG).
- **Región codificante:** Es la secuencia del gen que se transcribe a ARN mensajero y contiene las instrucciones para sintetizar la proteína. Comienza en el codón de inicio (ATG en el ADN codificante / AUG en el ARN) y se lee en tripletes sucesivos que codifican los aminoácidos. (En este caso, corresponde a la secuencia dada: ATG CCT GAA TGC).


### Extensión con Biopython:
#### Crea un script que lea un archivo FASTA con ADN y produzca la secuencia de ARNm. Experimenta cambiando la orientación de la hebra y observa qué ocurre.

Se descargó del NCBI (National Library of Medicine) un archivo FASTA correspondiente a la secuencia codificante (ADNc) del gen de la insulina humana [(Homo sapiens INS, variante 1, `NM_000207.3`)](https://www.ncbi.nlm.nih.gov/nuccore/NM_000207.3/). A continuación, se procesa la secuencia con Biopython para generar su transcrito de ARNm y analizar qué ocurre al invertir la orientación de la hebra.

````python
registro = SeqIO.read("sequence.fasta", "fasta")
adn = registro.seq

arnm_directo = adn.transcribe()

arnm_invertido = adn.reverse_complement().transcribe()

print("ARNm directo (primeras bases): ", arnm_directo[:40])
print("ARNm invertido (primeras bases): ", arnm_invertido[:40])
````

Al modificar la orientación de la hebra mediante .reverse_complement() (que invierte el sentido de lectura a la vez que cambia cada base por su complementaria):

- El transcrito de ARN resultante cambia respecto al original. La pauta de lectura se invierte y se genera una combinación de codones totalmente distinta.

- Se destruyen las señales biológicas esenciales del gen original (como la región reguladora inicial y el codón de inicio AUG). Si este transcrito invertido se intentara traducir en el ribosoma, daría lugar a una proteína completamente distinta, no funcional o abortada prematuramente por codones de parada fuera de lugar.

En conclusión, la transcripción se debe realizar obligatoriamente a partir de la hebra molde (dirección $3' \to 5'$) correcta para obtener un ARNm funcional que codifique la proteína deseada.

## Ejercicio 3. Traducción del ARNm a proteína 


### 1. Usa el siguiente transcrito de ARN: 5’ – AUG UAU GCU UAA – 3’.

A partir de la secuencia de transcrito de ARN proporcionada en el enunciado, se identifican las siguientes señales de traducción:
- **Codón de inicio:** Corresponde al triplete AUG, situado en el extremo 5' de la secuencia.
- **Codón de paro:** Corresponde al triplete UAA, situado en el extremo 3' de la secuencia.

### 3. Traduce la secuencia en una cadena de aminoácidos.
Aplicando el código genético a cada codón del transcrito:
- **AUG**: Codifica para el aminoácido Metionina (Met).
- **UAU**: Codifica para el aminoácido Tirosina (Tyr).
- **GCU**: Codifica para el aminoácido Alanina (Ala).
- **UAA**: Actúa como codón de paro, señalizando la terminación de la síntesis proteica, por lo que no añade ningún aminoácido a la cadena.

La secuencia traducida genera la siguiente cadena de aminoácidos: Met - Tyr - Ala.

### 4. Reflexiona: ¿qué pasaría si el codón de inicio mutara de AUG a GUG? ¿Qué ocurriría si el codón de paro desapareciera por mutación?
Si se produjeran estas alteraciones en la secuencia del ARN, las consecuencias directas sobre la síntesis proteica serían las siguientes:

- **Mutación del codón de inicio (de AUG a GUG)**: El ribosoma no reconocería el punto de partida de la traducción. Esto provocaría que la proteína no se sintetice en absoluto o que, si existe otro triplete de inicio más adelante, se genere una proteína truncada (más corta) que perdería su función original.

- **Desaparición del codón de paro:** El ribosoma no recibiría la señal de finalización y continuaría leyendo y añadiendo aminoácidos hasta encontrar el siguiente codón de parada. Esto produciría una proteína anormalmente larga, alterando su estructura y volviéndola no funcional.

### Extensión con Biopython:
#### Utiliza el módulo Bio.Seq para traducir automáticamente el ARNm a una secuencia proteica y  comprueba si el resultado coincide con el tuyo.

````python
secuencia_arn = Seq("AUGUAUGCUUAA")

secuencia_proteica = secuencia_arn.translate()
secuencia_proteica_limpia = secuencia_arn.translate(to_stop=True)

print(f"Secuencia de ARNm: {secuencia_arn}")
print(f"Cadena proteica (con stop '*'): {secuencia_proteica}")
print(f"Cadena proteica (hasta stop):  {secuencia_proteica_limpia}")
````

## Ejercicio 4. Splicing alternativo

### 1. Considera un gen con 5 exones: Exón 1 – Exón 2 – Exón 3 – Exón 4 – Exón 5.

### 2. Diseña al menos dos combinaciones de splicing alternativo (por ejemplo, 1-2-4-5 o 1-3-5).

A partir del pre-ARNm (Exón 1 – Exón 2 – Exón 3 – Exón 4 – Exón 5), se obtienen las siguientes variantes:
- **Isoforma A:** Exón 1 – Exón 2 – Exón 3 – Exón 5 (se excluye el exón 4).
- **Isoforma B:** Exón 1 – Exón 3 – Exón 5 (se excluyen los exones 2 y 4).

**Regla biológica**: Cualquier combinación es válida siempre que se respete el orden secuencial $5' \to 3'$ (nunca se permutan exones, como por ejemplo un hipotético 3-2-1).

### 3. Explica qué diferencias esperarías en las proteínas resultantes.

Las isoformas generadas presentarán variaciones clave debido a la inclusión o exclusión de exones:

- **Longitud y estructura tridimensional:** Al variar el número de exones traducidos, cambia la secuencia de aminoácidos y el tamaño del polipéptido, lo que modifica directamente su plegamiento espacial.

- **Pérdida o ganancia de dominios funcionales:** Los exones suelen corresponderse con dominios proteicos específicos. La omisión de un exón concreto puede eliminar un sitio de anclaje a membrana, un dominio catalítico o una región de unión a ligandos.

- **Regulación y localización celular:** Pequeños cambios en la secuencia pueden alterar la afinidad por sustratos, la vida media de la proteína o su localización.

### 4. Reflexiona: ¿por qué este mecanismo aumenta la diversidad proteica sin necesidad de más genes?

El splicing alternativo rompe la correspondencia "un gen = una proteína". A partir de una única secuencia codificante en el ADN, la selección combinatoria de exones genera múltiples transcritos maduros (isoformas de ARNm), cada uno traducible a una variante proteica distinta.

Esto permite maximizar la capacidad funcional y la complejidad del proteoma sin incrementar el tamaño del genoma ni requerir un gen individual para cada función celular.

## Ejercicio 5. Introducción a las proteínas

### 1. Considera la siguiente secuencia de aminoácidos: Met – Ile – Ser – Gly – Val – Lys – His.

### 2. Identifica el extremo N y el extremo C de la cadena.
Dada la secuencia anterior:

- **Extremo N (N-terminal):** Corresponde a la Metionina (Met), situada al inicio (izquierda). Es el primer elemento de la secuencia donde queda libre el grupo químico amino (con nitrógeno).
- **Extremo C (C-terminal):** Corresponde a la Histidina (His), situada al final (derecha). Es el último elemento de la secuencia donde queda libre el grupo químico carboxilo (con carbono).

Por convenio biológico, las proteínas siempre se leen y se construyen en esta dirección, desde el inicio (extremo N) hasta el final (extremo C).

### 3. Reflexiona: ¿cómo influye el orden de los aminoácidos en la estructura final de la proteína? ¿Qué ocurriría si hubiera una mutación que cambiara un aminoácido hidrofóbico por uno hidrofílico en una región interna de la proteína?

- **Influencia del orden de los aminoácidos:** La secuencia de aminoácidos determina cómo se pliega la cadena en el espacio para adoptar su forma tridimensional funcional. Cada aminoácido tiene propiedades físicas y químicas distintas; por tanto, el orden en el que aparecen dicta qué interacciones pueden formarse a lo largo de la molécula para mantener su estructura estable.

- **Impacto de cambiar un aminoácido hidrofóbico por uno hidrofílico en el interior:** Las proteínas solubles suelen plegarse situando los aminoácidos hidrofóbicos (que repelen el agua) en su núcleo interno, dejando los hidrofílicos (afines al agua) hacia el exterior. Introducir un aminoácido hidrofílico en el centro genera inestabilidad porque esta región carece de agua para interactuar con él. Como consecuencia, la proteína puede plegarse incorrectamente, perder su forma y quedar total o parcialmente no funcional.

Para responder a esta pregunta conviene separar dos ideas: dónde ocurren más errores y dónde un error puede tener consecuencias más graves.

Sin embargo, sus efectos suelen ser temporales, ya que la célula produce múltiples copias de ARNm y proteínas, y dispone de mecanismos para degradar moléculas defectuosas.

* **El punto más crítico (Replicación del ADN):** Un error durante la replicación que no sea corregido puede convertirse en una mutación permanente en el ADN. En ese caso, la información genética del gen queda modificada y puede dar lugar a ARNm y proteínas alteradas cada vez que ese gen sea expresado. Además, la mutación puede transmitirse a las células descendientes cuando la célula se divide.


### 6. Programa un pipeline que realice los tres procesos (replicación, transcripción, y traducción.

````python
def pipeline_dogma_central(fasta_file: str):
    registro = SeqIO.read(fasta_file, "fasta")

    # La secuencia del FASTA corresponde al ARNm, por lo que
    # se convierte T/U para trabajar con una secuencia de ADN.
    adn = Seq(str(registro.seq).replace("U", "T"))

    # 1. Replicación
    hebra_cmplment = adn.complement()

    # 2. Transcripción
    arnm = adn.transcribe()

    # 3. Traducción
    inicio = arnm.find("AUG")

    if inicio != -1:
        orf = arnm[inicio:]
        resto = len(orf) % 3

        if resto != 0:
            orf = orf[:-resto]

        proteina = orf.translate(to_stop=True)
    else:
        proteina = Seq("")

    print(f"PIPELINE DOGMA CENTRAL | ID: {registro.id} ({len(adn)} nt)\n")
    print(f"1. Replicación (hebra complementaria):\n{hebra_cmplment[:60]}...\n")
    print(f"2. Transcripción (ARNm 5'->3'):\n{arnm[:60]}...\n")
    print(f"3. Traducción ({len(proteina)} aa):\n{proteina}")


pipeline_dogma_central("sequence.fasta")
````

````python
registro = SeqIO.read("sequence.fasta", "fasta")
secuencia = str(registro.seq).replace("U", "T")
adn = Seq(secuencia)

print(f"Iniciando pipeline... Secuencia de {len(adn)} nucleótidos cargada.")

# 1. Replicación
hebra_comp = adn.complement()
print(f"1. Replicación completada. Hebra complementaria generada ({len(hebra_comp)} nt).")

# 2. Transcripción
arnm = adn.transcribe()
print(f"2. Transcripción completada. ARNm sintetizado ({len(arnm)} nt).")

# 3. Traducción
inicio = arnm.find("AUG")
if inicio != -1:
    print(f" -> Codón de inicio 'AUG' detectado en la posición {inicio}.")
    orf = arnm[inicio:]

    resto = len(orf) % 3
    if resto != 0:
        orf = orf[:-resto]
        print(f" -> Ajustando marco de lectura: se omitieron {resto} nucleótidos sobrantes.")

    proteina = orf.translate(to_stop=True)
    print(f"3. Traducción completada. Se obtuvo una proteína de {len(proteina)} aminoácidos.")
    print(f"Proteína final:\n{proteina}")
else:
    print("3. Traducción abortada: No se detectó codón de inicio 'AUG'.")
````
